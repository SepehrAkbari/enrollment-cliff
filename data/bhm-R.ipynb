```r
knitr::opts_chunk$set(echo = TRUE)
library(tidyverse)
library(brms)
library(tidybayes)
library(bayesplot)
library(janitor)
library(here)
set.seed(123)
```

```r
# Load data
data <- read_csv(here("data", "full_data.csv"))
glimpse(data)
```

```r
data <- data %>%
  mutate(
    state = as.factor(state),
    reason_closure = as.factor(reasonClosure)
  )

# example, colleges closed for mutual benefit reason per state
data %>%
  filter(reason_closure == "mutual_benefit") %>%
  group_by(state) %>%
  summarise(count = n()) %>%
  ggplot(aes(x = state, y = count)) +
  geom_col() +
  labs(title = "Colleges Closed for Mutual Benefit by State",
       x = "State",
       y = "Number of Colleges") +
  theme_minimal() +
  coord_flip()
```
