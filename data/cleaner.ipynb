```python
endow = pd.read_csv("sepehr_endowments.csv")

endow['hasMissing'] = endow['hasMissing'].replace({0: False, 1: True})

endow['index'] = endow['index'].astype('Int64')
endow['yearClosed'] = endow['yearClosed'].astype('Int64')

endow.loc[endow['hasMissing'] == True, endow.columns[1:]] = endow.loc[endow['hasMissing'] == True, endow.columns[1:]].infer_objects().replace(0, np.nan)

#drop row with no index
endow = endow.dropna(subset=['index'])

# if none of the columns that start with 'Endowment' have a value, then mark noData as True
endow['noData'] = endow[[x for x in endow.columns if x.startswith('Endowment')]].isnull().all(axis=1)
```
