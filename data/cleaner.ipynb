```python
rev = pd.read_excel("forStateStuff.xlsx")

for year in range(2019, 2025):
    for state in rev['state']:
        rev['fourYearAppropriations' + str(year)] = # the "Four-Year Education Appropriations" column for that year.
        rev['twoYearAppropriations' + str(year)] = # the "Two-Year Education Appropriations" column for that year.
        rev['fourYearNetFTE' + str(year)] = # the "Four-Year Net FTE Enrollment" column for that year.
        rev['twoYearNetFTE' + str(year)] = # the "Two-Year Net FTE Enrollment" column for that year.
        rev['fourYearNetTuitionAndFeeRevenue' + str(year)] = # the "Four-Year Net Tuition and Fee Revenue" column for that year.
        rev['twoYearNetTuitionAndFeeRevenue' + str(year)] = # the "Two-Year Net Tuition and Fee Revenue" column for that year.
        rev['fourYearTotalEducationRevenue' + str(year)] = # the "Four-Year Total Education Revenue" column for that year.
        rev['twoYearTotalEducationRevenue' + str(year)] = # the "Two-Year Total Education Revenue" column for that year.
```

```python

result = rev[['state']].copy()

columns_to_extract = {
    'Four-Year Education Appropriations': 'fourYearAppropriations',
    'Two-Year Education Appropriations': 'twoYearAppropriations',
    'Four-Year Net FTE Enrollment': 'fourYearNetFTE',
    'Two-Year Net FTE Enrollment': 'twoYearNetFTE',
    'Four-Year Net Tuition and Fee Revenue': 'fourYearNetTuitionAndFeeRevenue',
    'Two-Year Net Tuition and Fee Revenue': 'twoYearNetTuitionAndFeeRevenue',
    'Four-Year Total Education Revenue': 'fourYearTotalEducationRevenue',
    'Two-Year Total Education Revenue': 'twoYearTotalEducationRevenue',
    'Two-Year Student Share': 'twoYearStudentShare',
    'Four-Year Student Share': 'fourYearStudentShare'
}
for year in range(2019, 2025):
    year_data = sheeo[sheeo['FY'] == year][['state'] + list(columns_to_extract.keys())]
    year_data = year_data.rename(columns={old: f"{new}{year}" for old, new in columns_to_extract.items()})
    result = result.merge(year_data, left_on='state', right_on='state', how='left')
    if 'State' in result.columns:
        result = result.drop('State', axis=1)
result.to_excel("cleaned_sheeo_data.xlsx", index=False)
```
