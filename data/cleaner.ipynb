```python
import pandas as pd
import numpy as np
import openpyxl
```

```python
df_college_closure = pd.read_excel("Bita_college_closure/college_closures.xlsx")
df_college_closure = df_college_closure.dropna(subset=['name'])
df_college_closure.head()
```

Converting UID and year established to integer

```python
for i in df_college_closure['UID']:
    if df_college_closure.notna()['UID'].any():
        df_college_closure['UID'] = df_college_closure['UID'].astype('Int64')

df_college_closure['established'] = df_college_closure['established'].astype('Int64')
```

Removing some unnecessary columns

```python
df_college_closure = df_college_closure.drop(columns=['address'])
```

Some renaming

```python
# renaming columns
df_college_closure = df_college_closure.rename(columns={
    'closure year': 'closed_year',
    'campus setting': 'location_type',
    'campus setting 2': 'location_type2'
})

df_college_closure.to_excel("college_closures_1.xlsx", index=False)
```
