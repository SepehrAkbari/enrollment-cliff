```python
import pandas as pd

df = pd.read_csv('data.csv', keep_default_na=False, na_values=[''])
# df.drop(columns=['totalByStateMean', 'totalByRegionMean', 'closedByStateMean', 'closedByRegionMean'], inplace=True)
df
```

```python
years = [2020, 2021, 2022, 2023, 2024, 2025]

# df['totalClosedByState'] = 0
# df['totalClosedByRegion'] = 0

# for year in years:
#     df['totalClosedByState'] += df[f'closedByState_{year}']
#     df[f'totalByState_{year}'] = df[f'totalByState_{year}'] - df['totalClosedByState']

#     df['totalClosedByRegion'] += df[f'closedByRegion_{year}']
#     df[f'totalByRegion_{year}'] = df[f'totalByRegion_{year}'] - df['totalClosedByRegion']

# df = df.drop(columns=['totalClosedByState'])
# df = df.drop(columns=['totalClosedByRegion'])

df['closedNextYearByState'] = 0
df['closedNextYearByRegion'] = 0

for year in years:
    df[f'totalByState_{year}'] = df[f'totalByState_{year}'] - df['closedNextYearByState']
    df['closedNextYearByState'] += df[f'closedByState_{year}']

    df[f'totalByRegion_{year}'] = df[f'totalByRegion_{year}'] - df['closedNextYearByRegion']
    df['closedNextYearByRegion'] += df[f'closedByRegion_{year}']

df = df.drop(columns=['closedNextYearByState'])
df = df.drop(columns=['closedNextYearByRegion'])

df
```

```python
df['totalByStateMean'] = df[['totalByState_2020', 'totalByState_2021', 'totalByState_2022', 'totalByState_2023']].mean(axis=1)
df['totalByRegionMean'] = df[['totalByRegion_2020', 'totalByRegion_2021', 'totalByRegion_2022', 'totalByRegion_2023']].mean(axis=1)
df['closedByStateMean'] = df[['closedByState_2020', 'closedByState_2021', 'closedByState_2022', 'closedByState_2023']].mean(axis=1)
df['closedByRegionMean'] = df[['closedByRegion_2020', 'closedByRegion_2021', 'closedByRegion_2022', 'closedByRegion_2023']].mean(axis=1)
df.to_csv('data.csv', index=False)
```
